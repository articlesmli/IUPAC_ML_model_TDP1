### Load and analyse PubChem BioAssay data <a name="1"></a>

```python
import pandas as pd 

# load the PubChem AID 686978 bioassy dataset
# https://pubchem.ncbi.nlm.nih.gov/bioassay/686978
df = pd.read_csv('AID_686978_data.csv', low_memory=False, on_bad_lines='skip')

# To avoid truncation of some columns during data frame display
pd.set_option('display.max_columns', None) 

# Display the data frame
print('Shape of the data frame: ', df.shape)
df.head()
```

```python
df.info()
```

```python
# Remove missing values
df = df[df['PUBCHEM_CID'].notna()]
df.shape
```

```python
# Remove duplicates without keeping a sample ofthem and reset the indexes 
df = df.drop_duplicates(subset='PUBCHEM_CID', keep=False).reset_index(drop=True)
df.shape
```

```python
# Turn CID float data type into integer
df['PUBCHEM_CID'] = df['PUBCHEM_CID'].astype('int64') 
```

```python
# Filter inhibitors 
df = df[['PUBCHEM_CID',
         'PUBCHEM_SID',
         'PUBCHEM_EXT_DATASOURCE_SMILES',
         'PUBCHEM_ACTIVITY_OUTCOME']] 
df.shape # 646675
```

```python
df.info()
```

## Create a data frame with targets <a name="2"></a>

```python
# Fix data type issue in the raw data
df['PUBCHEM_ACTIVITY_OUTCOME'] = df['PUBCHEM_ACTIVITY_OUTCOME'].astype(str) 
# Get unique values from 'column1'
unique_values = df['PUBCHEM_ACTIVITY_OUTCOME'].unique()
unique_values 
```

```python
# Create a dictionary with the desired column names
df.rename(columns={'PUBCHEM_CID':'CID',
                   'PUBCHEM_SID':'SID',
                   'PUBCHEM_EXT_DATASOURCE_SMILES':'SMILES',
                   'PUBCHEM_ACTIVITY_OUTCOME':'target'}, inplace=True)

# Set the option to explicitly handle downcasting
pd.set_option('future.no_silent_downcasting', True)

# Create a mapping dictionary to replace string values with numeric values
mapping = {'Active': 1, 'Inactive': 0, 'Inconclusive':2}

# Replace string values with numeric values using the mapping dictionary
df['target'] = df['target'].replace(mapping)

# Display the data frame
print('Shape of the data frame: ', df.shape)
df.head()
```

```python
# Turn CID float data type into integer
df['target'] = pd.to_numeric(df['target']) 
```

```python
# Class distribution
df['target'].value_counts()
```

```python
# Dataframe shape
df_1 = df[df['target']==1]
df_1.shape
```

## Asign PAIN flags  <a name="3"></a>

```python
import pandas as pd
from rdkit import Chem
from rdkit.Chem.FilterCatalog import FilterCatalog, FilterCatalogParams

# Define the PAINS Filter Source
# create a filter catalog for Pan Assay Interference Substructures (PAINS).
params = FilterCatalogParams()
params.AddCatalog(FilterCatalogParams.FilterCatalogs.PAINS)
pains_catalog = FilterCatalog(params)

def check_pains(smiles):
    """
    Checks a given SMILES string against the RDKit PAINS filter catalog.
    
    Returns:
        bool: True if the compound is flagged as a PAIN, False otherwise.
    """
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return False  # Cannot process invalid SMILES
    
    # Run the molecule against the PAINS catalog
    is_pain = pains_catalog.HasMatch(mol)
    return is_pain


# Apply the RDKit check to generate the actual PAIN flag
# Create a new boolean column: True if it is a PAIN, False otherwise.
df['PAIN_flag'] = df['SMILES'].apply(check_pains)

# Convert the boolean flag to 'PAIN' / 'No' for clearer output
df['PAIN_flag'] = df['PAIN_flag'].apply(lambda x: 'PAIN' if x else 'No')

print("--- Original DataFrame with RDKit PAINS Check Results ---")
print(df)
print("-" * 70)
print(f"Total compounds: {len(df)}")
print(f"PAIN flagged compounds: {df['PAIN_flag'].eq('PAIN').sum()}")
print("-" * 70)
```

```python
# Class distribution
df['target'].value_counts()
```

```python
# Create a csv file with flagged samples
df.to_csv('HF_TDP1_inhibitors_with_PAIN_flags.csv', index=False)
```

 ## Remove the PAIN samples <a name="4"></a>

```python
# Define the condition for samples (rows) to KEEP
#'PAIN_flag' column is NOT equal to 'PAIN'.
condition_to_keep = df['PAIN_flag'] == 'No'

# Apply the Boolean mask to filter the DataFrame
df_filtered_samples = df[condition_to_keep].copy()

print("\n--- Filtered DataFrame (PAIN Samples Excluded) ---")
print(df_filtered_samples)
print("-" * 70)
```

```python
# Class distribution 
df_filtered_samples['target'].value_counts()
```

```python
# Chek again for PAINs
df_filtered_samples['PAIN_flag'] = df_filtered_samples['SMILES'].apply(check_pains)

# Convert the boolean flag to 'PAIN' / 'No' for clearer output
df_filtered_samples['PAIN_flag'] = df_filtered_samples['PAIN_flag'].apply(lambda x: 'PAIN' if x else 'No')

print("--- Original DataFrame with RDKit PAINS Check Results ---")
print(df_filtered_samples)
print("-" * 70)
print(f"Total compounds: {len(df_filtered_samples)}")
print(f"PAIN flagged compounds: {df_filtered_samples['PAIN_flag'].eq('PAIN').sum()}")
print("-" * 70)
```

```python
# Display the data frame
print('Shape of df_filtered_samples: ', df_filtered_samples.shape)
df_filtered_samples.head()
```

```python
# Class ditribution 
df_filtered_samples['target'].value_counts()
```

```python
# Create csv file without PAINs
df_filtered_samples.to_csv('HF_TDP1_inhibitors_no_PAINs.csv', index=False)
```

## Reduce the inactive compounds <a name="5"></a>

```python
# Load data from PubChem BioAssay AID 1996 
df_solubility = pd.read_csv('AID_1996_data.csv')

# Display the data frame
print('Shape of df_solubility: ', df_solubility.shape)
df_solubility.head()
```

```python
df_solubility.rename(columns={'PUBCHEM_EXT_DATASOURCE_SMILES':'SMILES'}, inplace=True)
```

```python
# Sift the main data frame with targets using the solubility data frame 
df = pd.merge(df, df_solubility, on=['SMILES', 'CID', 'SID'])

df = df[['CID', 'SID', 'SMILES', 'target']]
```

```python
# Display the data frame
print('Shape of df: ', df.shape)
df.head()
```
